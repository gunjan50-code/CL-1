# ML Experiment 6 (A6) - Reinforcement Learning
### Option A: Q-Learning agent exploring a maze environment

**Aim:** Implement reinforcement learning using a maze environment that the agent must explore to reach the goal, using the Q-learning algorithm.

**Step 1:** Import libraries and set up the maze environment (a 4x4 grid: start at 0, goal at 15).

```python
import numpy as np

size = 4
n_states = size * size
goal = n_states - 1
n_actions = 4   # 0=up, 1=down, 2=left, 3=right

def step(state, action):
    r, c = divmod(state, size)
    if action == 0 and r > 0: r -= 1
    elif action == 1 and r < size-1: r += 1
    elif action == 2 and c > 0: c -= 1
    elif action == 3 and c < size-1: c += 1
    new_state = r * size + c
    reward = 10 if new_state == goal else -1
    return new_state, reward

print("Maze ready:", n_states, "states, goal at", goal)
```

**Step 2:** Build the reinforcement learning model - initialize the Q-table and set the learning parameters.

```python
Q = np.zeros((n_states, n_actions))
alpha = 0.1    # learning rate
gamma = 0.9    # discount factor
epsilon = 0.2  # exploration rate
print("Q-table shape:", Q.shape)
```

**Step 3:** Train the model over many episodes using the Q-learning update rule.

```python
np.random.seed(42)
for episode in range(500):
    state = 0
    for _ in range(100):
        if np.random.rand() < epsilon:
            action = np.random.randint(n_actions)
        else:
            action = np.argmax(Q[state])
        new_state, reward = step(state, action)
        Q[state, action] += alpha * (reward + gamma * np.max(Q[new_state]) - Q[state, action])
        state = new_state
        if state == goal:
            break
print("Training complete over 500 episodes")
```

**Step 4:** Test the model - follow the learned policy (best action in each state) from start to goal.

```python
state = 0
path = [0]
for _ in range(20):
    action = np.argmax(Q[state])
    state, _ = step(state, action)
    path.append(state)
    if state == goal:
        break
print("Learned path from start to goal:", path)
print("Steps taken:", len(path) - 1)
```
