# Practical 11 (B6) - Data Aggregation
### Analyzing Sales Performance by Region in a Retail Company

**Task 1:** Import the "Retail_Sales_Data.csv" dataset.

(Sample sales data is generated here to represent the dataset.)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)
regions = ["North", "South", "East", "West"]
categories = ["Electronics", "Clothing", "Grocery", "Toys"]
n = 200
df = pd.DataFrame({
    "Date": pd.date_range("2024-01-01", periods=n, freq="D").strftime("%Y-%m-%d"),
    "Region": np.random.choice(regions, n),
    "Category": np.random.choice(categories, n),
    "Quantity": np.random.randint(1, 20, n),
    "SalesAmount": np.random.randint(100, 5000, n)
})
df.head()
```

**Task 2:** Explore the dataset to understand its structure and content.

```python
print("Shape:", df.shape)
print("Columns:", list(df.columns))
df.describe()
```

**Task 3:** Identify the relevant variables for aggregating sales data (Region, Category, SalesAmount).

```python
print("Grouping variables: Region, Category")
print("Measure to aggregate: SalesAmount")
```

**Task 4:** Group the sales data by region and calculate the total sales amount for each region.

```python
region_sales = df.groupby("Region")["SalesAmount"].sum()
print(region_sales)
```

**Task 5:** Create a bar plot and a pie chart to visualize the sales distribution by region.

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
region_sales.plot(kind="bar", color="teal", ax=axes[0], title="Sales by Region")
axes[0].set_ylabel("Total Sales")
region_sales.plot(kind="pie", autopct="%1.1f%%", ax=axes[1], title="Sales Share by Region")
axes[1].set_ylabel("")
plt.tight_layout(); plt.show()
```

**Task 6:** Identify the top-performing region based on the highest sales amount.

```python
top_region = region_sales.idxmax()
print("Top-performing region:", top_region, "with total sales", region_sales.max())
```

**Task 7:** Group the sales data by region and product category to calculate the total for each combination.

```python
pivot = df.pivot_table(values="SalesAmount", index="Region", columns="Category", aggfunc="sum")
pivot
```

**Task 8:** Create a grouped/stacked bar plot to compare sales across regions and product categories.

```python
pivot.plot(kind="bar", stacked=True, figsize=(10, 5), colormap="viridis")
plt.title("Sales by Region and Product Category")
plt.xlabel("Region"); plt.ylabel("Total Sales")
plt.legend(title="Category"); plt.tight_layout(); plt.show()
```
