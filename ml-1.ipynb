# ML Experiment 1 (A1) - Feature Transformation
### Option A: PCA Algorithm for Dimensionality Reduction on Wine Dataset

**Aim:** Apply the PCA algorithm to the Wine dataset so that most of the variation in the measurements is captured by a small number of principal components, making it easier to distinguish between the wine classes.

**Step 1:** Import the required libraries.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

sns.set(style="whitegrid")
print("Libraries ready")
```

**Step 2:** Import the Wine.csv dataset.

```python
url = "Wine.csv"
df = pd.read_csv(url)
df.head()
```

**Step 3:** Explore the dataset and check for missing values.

```python
print("Shape (rows, columns):", df.shape)
print("\nColumn names:", list(df.columns))
print("\nWine classes in Customer_Segment:", df["Customer_Segment"].unique())
print("\nMissing values:")
print(df.isnull().sum())
```

**Step 4:** Separate the features (X) from the target class label (y).

```python
X = df.drop("Customer_Segment", axis=1)
y = df["Customer_Segment"]

print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

**Step 5:** Standardize the features (mean = 0, standard deviation = 1). PCA requires scaled data so that every variable contributes fairly.

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_scaled_df = pd.DataFrame(X_scaled, columns=X.columns)
X_scaled_df.head()
```

**Step 6:** Apply the PCA algorithm and check how much variance each principal component captures.

```python
pca = PCA()
X_pca = pca.fit_transform(X_scaled)

explained = pca.explained_variance_ratio_
for i, val in enumerate(explained):
    print(f"PC{i+1}: {val*100:.2f}% variance")
```

**Step 7:** Plot the cumulative explained variance to decide how many principal components to keep.

```python
cumulative = np.cumsum(explained)

plt.figure(figsize=(8, 4))
plt.plot(range(1, len(cumulative)+1), cumulative, marker="o", color="blue")
plt.axhline(y=0.95, color="red", linestyle="--", label="95% variance")
plt.title("Cumulative Explained Variance by Principal Components")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Variance Explained")
plt.legend(); plt.tight_layout(); plt.show()
```

**Step 8:** Transform the data into just 2 principal components (reducing 13 variables down to 2).

```python
pca2 = PCA(n_components=2)
X_2d = pca2.fit_transform(X_scaled)

pca_df = pd.DataFrame(X_2d, columns=["PC1", "PC2"])
pca_df["Wine_Class"] = y.values

print("Variance captured by 2 components:", round(pca2.explained_variance_ratio_.sum()*100, 2), "%")
pca_df.head()
```

**Step 9:** Visualize the two principal components. Each colour is a different wine class - PCA makes them easy to distinguish in 2D.

```python
plt.figure(figsize=(8, 6))
sns.scatterplot(data=pca_df, x="PC1", y="PC2", hue="Wine_Class", palette="Set1", s=80)
plt.title("Wine Classes Separated by 2 Principal Components")
plt.xlabel("Principal Component 1")
plt.ylabel("Principal Component 2")
plt.legend(title="Wine Class")
plt.tight_layout(); plt.show()
```
