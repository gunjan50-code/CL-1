### Analyzing Air Quality Index (AQI) Trends in a City

**Task 1:** Import the "City_Air_Quality.csv" dataset.

(Sample AQI data is generated here to represent the dataset.)

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(1)
dates = pd.date_range("2024-01-01", periods=30, freq="D")
df = pd.DataFrame({
    "Date": dates,
    "PM2_5": np.random.randint(30, 120, 30),
    "PM10": np.random.randint(40, 160, 30),
    "CO": np.round(np.random.uniform(0.3, 2.0, 30), 2)
})
df["AQI"] = (df["PM2_5"]*0.5 + df["PM10"]*0.3 + df["CO"]*20).astype(int)
df.head()
```

**Task 2:** Explore the dataset to understand its structure and content.

```python
print("Shape:", df.shape)
print("Columns:", list(df.columns))
df.describe()
```

**Task 3:** Identify the relevant variables for visualizing AQI trends (date, pollutant levels, AQI).

```python
print("Time variable : Date")
print("Pollutants    : PM2_5, PM10, CO")
print("Target metric : AQI")
```

**Task 4:** Create a line plot to visualize the overall AQI trend over time.

```python
plt.figure(figsize=(10, 4))
plt.plot(df["Date"], df["AQI"], marker="o", color="crimson")
plt.title("Overall AQI Trend Over Time")
plt.xlabel("Date"); plt.ylabel("AQI")
plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

**Task 5:** Plot individual pollutant levels on separate line plots to visualize their trends.

```python
fig, axes = plt.subplots(3, 1, figsize=(10, 8), sharex=True)
axes[0].plot(df["Date"], df["PM2_5"], color="blue");  axes[0].set_ylabel("PM2.5")
axes[1].plot(df["Date"], df["PM10"], color="green");  axes[1].set_ylabel("PM10")
axes[2].plot(df["Date"], df["CO"], color="orange");   axes[2].set_ylabel("CO")
axes[0].set_title("Individual Pollutant Trends")
plt.xlabel("Date"); plt.xticks(rotation=45); plt.tight_layout(); plt.show()
```

**Task 6:** Use a bar plot to compare AQI values across different dates.

```python
plt.figure(figsize=(10, 4))
plt.bar(df["Date"].dt.strftime("%m-%d"), df["AQI"], color="skyblue")
plt.title("AQI Comparison Across Dates")
plt.xlabel("Date"); plt.ylabel("AQI")
plt.xticks(rotation=90); plt.tight_layout(); plt.show()
```

**Task 7:** Create box plots to analyze the distribution of each pollutant.

```python
plt.figure(figsize=(7, 5))
df[["PM2_5", "PM10", "CO"]].plot(kind="box")
plt.title("Distribution of Pollutant Levels")
plt.ylabel("Value")
plt.show()
```

**Task 8:** Use a scatter plot to explore the relationship between AQI and a pollutant (PM2.5).

```python
plt.figure(figsize=(7, 5))
plt.scatter(df["PM2_5"], df["AQI"], color="purple")
plt.title("AQI vs PM2.5")
plt.xlabel("PM2.5"); plt.ylabel("AQI")
plt.grid(True); plt.show()
```
